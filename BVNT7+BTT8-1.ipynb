### W7A16
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
arr = list(map(int, input().split()))

counts = {}
for x in arr:
    counts[x] = counts.get(x, 0) + 1

removes = 0
for x, count in counts.items():
    if x == 0:
        removes += count
    elif count >= x:
        removes += count - x
    else:
        removes += count

print(removes)
```

### W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.  

```python
arr = list(map(int, input().split()))

candidate = None
count = 0

for num in arr:
    if count == 0:
        candidate = num
    if num == candidate:
        count += 1
    else:
        count -= 1

actual_count = 0
for num in arr:
    if num == candidate:
        actual_count += 1

if actual_count > len(arr) // 2:
    print(candidate)
else:
    print("Not found")
```

### W7A18
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng, lẻ giảm dần


```python
arr = list(map(int, input().split()))

evens = []
odds = []

for num in arr:
    if num % 2 == 0:
        evens.append(num)
    else:
        odds.append(num)

evens.sort()
odds.sort(reverse=True)

result = evens + odds
print(*(result))
```

### W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào


```python
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))

is_descending = False
if len(arr1) > 1 and arr1[0] > arr1[-1]:
    is_descending = True
elif len(arr2) > 1 and arr2[0] > arr2[-1]:
    is_descending = True

result = []
i = 0
j = 0

while i < len(arr1) and j < len(arr2):
    if is_descending:
        condition = arr1[i] > arr2[j]
    else:
        condition = arr1[i] < arr2[j]

    if condition:
        result.append(arr1[i])
        i += 1
    else:
        result.append(arr2[j])
        j += 1

while i < len(arr1):
    result.append(arr1[i])
    i += 1

while j < len(arr2):
    result.append(arr2[j])
    j += 1

print(*(result))
```

### W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
names = []
for _ in range(n):
    names.append(input().strip())

counts = {}
for name in names:
    counts[name] = counts.get(name, 0) + 1

best_name = ""
max_count = 0

for name in counts:
    if counts[name] > max_count:
        max_count = counts[name]
        best_name = name
    elif counts[name] == max_count:
        if name < best_name or best_name == "":
            best_name = name

print(best_name)
```

### W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
s = input().strip()
k = int(input())

to_remove = len(s) - k
stack = []

for digit in s:
    while to_remove > 0 and stack and stack[-1] < digit:
        stack.pop()
        to_remove -= 1
    stack.append(digit)

result = "".join(stack[:k])
print(result)
```

### W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s = input().strip()

stack = []
mapping = {")": "(", "}": "{", "]": "["}
valid = True

for char in s:
    if char in mapping.values():
        stack.append(char)
    elif char in mapping.keys():
        if not stack or stack[-1] != mapping[char]:
            valid = False
            break
        stack.pop()

if stack:
    valid = False

if valid:
    print("Hợp lệ")
else:
    print("Không hợp lệ")
```

### W7A28
Bà
̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
n = int(input())

limit = 2
while True:
    primes = [True] * limit
    primes[0] = primes[1] = False
    for i in range(2, int(limit**0.5) + 1):
        if primes[i]:
            for j in range(i*i, limit, i):
                primes[j] = False
                
    prime_list = [i for i, p in enumerate(primes) if p]
    
    max_gap = 0
    start_num = -1
    
    for i in range(len(prime_list) - 1):
        gap = prime_list[i+1] - prime_list[i] - 1
        if gap > max_gap:
            max_gap = gap
            start_num = prime_list[i] + 1
            
    if max_gap >= n - 1:
        result = list(range(start_num, start_num + max_gap))
        print(*(result[:n-1]))
        break
    limit *= 2
```

### W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau


```python
arr = list(map(int, input().split()))

if not arr:
    print("")
else:
    max_len = 1
    max_start = 0
    
    current_len = 1
    current_start = 0
    
    for i in range(1, len(arr)):
        if arr[i] > arr[i - 1]:
            current_len += 1
        else:
            if current_len > max_len:
                max_len = current_len
                max_start = current_start
            current_len = 1
            current_start = i
            
    if current_len > max_len:
        max_len = current_len
        max_start = current_start

    result = arr[max_start : max_start + max_len]
    print(*(result))
```

### W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
arr = list(map(int, input().split()))

if not arr:
    print("")
else:
    n = len(arr)
    L = [1] * n
    T = [-1] * n

    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j] and L[j] + 1 > L[i]:
                L[i] = L[j] + 1
                T[i] = j

    max_len = max(L)
    max_idx = L.index(max_len)

    result = []
    while max_idx != -1:
        result.append(arr[max_idx])
        max_idx = T[max_idx]

    result.reverse()
    print(*(result))
```

### W7A31
đổi chỗ đúng 2 chữ số để được số nhỏ nhất

```python
s = list(input().strip())
n = len(s)

swapped = False

for i in range(n):
    min_idx = i
    for j in range(n - 1, i, -1):
        if s[j] < s[min_idx]:
            if i == 0 and s[j] == '0':
                continue
            if min_idx == i or s[j] < s[min_idx]:
                min_idx = j
                
    if min_idx != i:
        s[i], s[min_idx] = s[min_idx], s[i]
        swapped = True
        break

print("".join(s))
```
