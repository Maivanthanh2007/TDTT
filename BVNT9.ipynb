Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển

### W9A17 
Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 


```python
import bisect


def length_of_lis(nums):
    if not nums:
        return 0

    tails = []
    for num in nums:
        idx = bisect.bisect_left(tails, num)
        if idx == len(tails):
            tails.append(num)
        else:
            tails[idx] = num

    return len(tails)


arr = list(map(int, input().split()))
print(length_of_lis(arr))
```

### W9A18
Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên
dương cho vào từ bàn phím) 


```python
def count_pairs_with_sum(arr, k):
    count = 0
    seen = {}

    for num in arr:
        complement = k - num
        if complement in seen:
            count += seen[complement]

        seen[num] = seen.get(num, 0) + 1

    return count


arr = list(map(int, input().split()))
k = int(input())

print(count_pairs_with_sum(arr, k))
```

### W9A19
Loại bỏ khỏi danh sách các phần tử trùng lặp, giữ lại đúng 1 lần xuất hiện, in lại theo đúng thứ tự
xuất hiện 


```python
def remove_duplicates(arr):
    seen = set()
    result = []
    for num in arr:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result


arr = list(map(int, input().split()))
print(*(remove_duplicates(arr)))
```

### W9A20
Thực hiện chuẩn hóa chuỗi kí tự:
a. bỏ khoảng trắng đầu cuối
b. các khoảng trắng giữa chỉ còn 1 dấu cách
c. viết hoa chữ cái đầu mỗi từ
KHÔNG DÙNG CÁC HÀM SẴN CỐ

```python
def normalize_string(s):
    n = len(s)

    start = 0
    while start < n and s[start] == " ":
        start += 1

    end = n - 1
    while end >= 0 and s[end] == " ":
        end -= 1

    if start > end:
        return ""

    result = []
    in_word = False

    for i in range(start, end + 1):
        char = s[i]

        if char != " ":
            if not in_word:
                if "a" <= char <= "z":
                    char = chr(ord(char) - 32)
                in_word = True
            else:
                if "A" <= char <= "Z":
                    char = chr(ord(char) + 32)

            result.append(char)
        else:
            if in_word:
                result.append(" ")
                in_word = False

    return "".join(result)


s = input()
print(normalize_string(s))
```

### W9A21
Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort) 


```python
def partition(arr, left, right):
    pivot = arr[right]
    i = left
    for j in range(left, right):
        if arr[j] > pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[right] = arr[right], arr[i]
    return i


def quickselect(arr, left, right, k):
    if left <= right:
        pivot_index = partition(arr, left, right)

        if pivot_index == k - 1:
            return arr[pivot_index]
        elif pivot_index > k - 1:
            return quickselect(arr, left, pivot_index - 1, k)
        else:
            return quickselect(arr, pivot_index + 1, right, k)
    return -1


arr = list(map(int, input().split()))
k = int(input())

print(quickselect(arr, 0, len(arr) - 1, k))
```

### W9A22
Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi
đầu vào có phải là anagram của nhau không

```python
from collections import Counter


def is_anagram(str1, str2):
    return Counter(str1) == Counter(str2)


str1 = input().strip()
str2 = input().strip()

if is_anagram(str1, str2):
    print("YES")
else:
    print("NO")
```

### W9A23
Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm) 


```python
def max_number_k_digits(s, k):
    to_remove = len(s) - k
    stack = []

    for digit in s:
        while to_remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            to_remove -= 1
        stack.append(digit)

    return "".join(stack[:k])


s = input().strip()
k = int(input())

print(max_number_k_digits(s, k))
```

### W9A24
Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5

```python
def compress_string(s):
    if not s:
        return ""

    result = []
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result.append(s[i - 1] + str(count))
            count = 1

    result.append(s[-1] + str(count))

    return "".join(result)


s = input().strip()
print(compress_string(s))
```

### W9A25
Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem
nó có phải là mật khẩu mạnh không. 

```python
def check_strong_password(password):
    if len(password) < 8:
        return False

    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False

    special_chars = {"!", "@", "#", "$"}

    for char in password:
        if "a" <= char <= "z":
            has_lower = True
        elif "A" <= char <= "Z":
            has_upper = True
        elif "0" <= char <= "9":
            has_digit = True
        elif char in special_chars:
            has_special = True

    return has_lower and has_upper and has_digit and has_special


password = input().strip()

if check_strong_password(password):
    print("STRONG")
else:
    print("WEAK")
```
